```python
df_clean["director"] = df_clean["director"].fillna("Unknown")
df_clean["cast"] = df_clean["cast"].fillna("Unknown")
df_clean["country"] = df_clean["country"].fillna("Unknown")
```

```python
df_clean["date_added"] = pd.to_datetime(
    df_clean["date_added"].str.strip(),
    format="%B %d, %Y",
    errors="coerce"
)
```

```python
content_counts = df_clean["type"].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(content_counts.index, content_counts.values)
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.title("Distribution of Movies and TV Shows")
plt.show()
```

```python
content_counts = df_clean["type"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    content_counts.values,
    labels=content_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Movies vs TV Shows Distribution")
plt.show()
```

```python
year_counts = df_clean["release_year"].value_counts().sort_index()

print(year_counts)
```

```python
year_counts.idxmax(), year_counts.max()
```

```python
year_counts = df_clean["release_year"].value_counts().sort_index()

plt.figure(figsize=(16, 6))

plt.plot(year_counts.index, year_counts.values, marker="o")

plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.title("Netflix Content Release Trends")

plt.xticks(year_counts.index, rotation=90)

plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
df_clean["listed_in"].head(10)
```

```python
genres = df_clean["listed_in"].str.split(", ")

genres.head()
```

```python
genre_list = genres.explode()

genre_list.head(10)
```

```python
genre_counts = genre_list.value_counts()

genre_counts.head(10)
```

```python
top_10_genres = genre_counts.head(10).sort_values()

plt.figure(figsize=(10, 6))

plt.barh(top_10_genres.index, top_10_genres.values)

plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.title("Top 10 Genres on Netflix")

plt.tight_layout()
plt.show()
```

```python
# df_clean["country"].head(10)
countries = df_clean["country"].str.split(", ")

# countries.head()
country_list = countries.explode()
country_counts = country_list.value_counts()

country_counts.head(10)
```

```python
top_countries = country_counts.drop("Unknown", errors="ignore").head(10)

top_countries = top_countries.sort_values()

plt.figure(figsize=(10, 6))

plt.barh(top_countries.index, top_countries.values)

plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.title("Top 10 Countries by Netflix Content")

plt.tight_layout()
plt.show()
```

```python
rating_counts = df_clean["rating"].value_counts().head(10)

rating_counts = rating_counts.sort_values()

plt.figure(figsize=(10, 6))

plt.barh(rating_counts.index, rating_counts.values)

plt.xlabel("Number of Titles")
plt.ylabel("Content Rating")
plt.title("Top 10 Netflix Content Ratings")

plt.tight_layout()
plt.show()
```

```python
movies = df_clean[df_clean["type"] == "Movie"]

movies["duration"].head(10)
```

```python
movies = df_clean[df_clean["type"] == "Movie"].copy()

movies["duration_mins"] = (
    movies["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)
movies["duration_mins"].describe()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=movies,
    x="duration_mins",
    bins=20,
    kde=True
)

plt.xlabel("Movie Duration (minutes)")
plt.ylabel("Number of Movies")
plt.title("Distribution of Movie Durations")
plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```
